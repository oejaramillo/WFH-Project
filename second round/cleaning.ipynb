```python
datos = pd.read_csv("classification_final.csv", index_col=0)
```

```python
datos.columns
```

```python
datos = datos[['classifier_id', 'ad_id', 'classification', 'ease', 'timestamp']]
```

```python
datos['timestamp'] = pd.to_datetime(datos['timestamp'], errors='coerce')
```

```python
# Ordenar el DataFrame por 'timestamp'
datos = datos.sort_values(by='timestamp')
```

```python
# Función para rellenar los valores vacíos con el timestamp anterior + un número aleatorio de minutos
def fill_missing_timestamps(row, previous_timestamp):
    if pd.isna(row['timestamp']):
        # Generar un número aleatorio de minutos entre 5 y 20
        random_minutes = np.random.randint(5, 21)
        # Retornar el timestamp anterior más los minutos aleatorios
        return previous_timestamp + timedelta(minutes=random_minutes)
    else:
        # Si el timestamp no está vacío, retornarlo como está
        return row['timestamp']
```

```python
# Inicializar el valor anterior como None
previous_timestamp = None

# Rellenar los valores vacíos
new_timestamps = []
for index, row in datos.iterrows():
    new_timestamp = fill_missing_timestamps(row, previous_timestamp)
    new_timestamps.append(new_timestamp)
    previous_timestamp = new_timestamp

# Asignar los nuevos timestamps al DataFrame
datos['timestamp'] = new_timestamps
```

```python
datos.to_csv("classification_secondRound_final.csv")
```
