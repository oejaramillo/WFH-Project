### Merging and counting tags accross rounds

```python
# Define a mapping
classification_mapping = {
    'tot_remote': 'WFH',
    'temp_remote': 'WFH_partial',
    'semi_remote': 'WFH_partial',
    'semi_presence': 'WFH_partial',
    'not_clear_remote': 'WFH_partial',
    'not_remote': 'Not WFH'
}

# Apply the mapping
second_round['classification'] = second_round['classification'].replace(classification_mapping)
```

```python
def disagreement_rate(df, item_col="ad_id"):  
    """
    df: dataframe with columns [item_col, classifier_id, classification]
    item_col: the column identifying the classified item (here: ad_id)
    returns: disagreement rate (fraction of items with disagreement)
    """
    disagreements = (
        df.groupby(item_col)["classification"]
          .nunique()
          .reset_index(name="n_unique_classifications")
    )
    
    disagreements["disagrees"] = disagreements["n_unique_classifications"] > 1
    
    disagreement_rate = disagreements["disagrees"].mean()
    
    return disagreement_rate, disagreements
```

```python
# Example:
first_rate, first_details = disagreement_rate(first_round, item_col="ad_id")
second_rate, second_details = disagreement_rate(second_round, item_col="ad_id")

print(f"First round disagreement rate: {first_rate:.2%}")
print(f"Second round disagreement rate: {second_rate:.2%}")
```
